```python
import requests
from time import sleep

# Define the URL to scrape (replace with the actual URL if needed)
url = "https://www.usdebtclock.org/"

def get_federal_debt():
  """Fetches the federal debt from the provided URL"""
  try:
    response = requests.get(url)
    response.raise_for_status()  # Raise exception for bad status codes
    # Parse the response to extract the debt value (implementation depends on website structure)
    # This example assumes the debt is within a specific element with ID "debt-value"
    debt_element = response.content.find(id="debt-value")
    print(debt_element)
    if debt_element:
      return debt_element.text.strip()
    else:
      print("Failed to find debt value element.")
      return None
  except requests.exceptions.RequestException as e:
    print(f"Error fetching data: {e}")
    return None

while True:
  # Fetch the debt
  debt = get_federal_debt()
  if debt:
    print(f"Current federal debt: {debt}")
  else:
    print("Failed to retrieve debt data.")
  # Wait for 2 minutes before next attempt
  sleep(120)
```
